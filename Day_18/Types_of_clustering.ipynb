## hierarchical clustering

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
import pandas as pd
import numpy as np
%matplotlib inline
import seaborn as sns

```

```python
from sklearn import datasets
```

```python
## Import IRIS Dataset

iris = datasets.load_iris()
```

```python
iris_data = pd.DataFrame(iris.data)
```

```python
iris_data
```

```python
iris_data.columns = iris.feature_names
```

```python
iris_data.columns
```

```python
X = pd.DataFrame(iris.data)
y = iris.target
```

```python
X
```

```python
X.head()
```

```python
y
```

```python
# Standardize the features

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(iris_data)
```

```python
X_scaled
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2)
```

```python
pca_scaled = pca.fit_transform(X_scaled)
```

```python
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
```

```python
plt.figure(figsize=(15, 7))
plt.title('Hierarchical Clustering Dendrogram (Iris Dataset)')
plt.xlabel('Sample Index')
plt.ylabel('Distance')
linked = linkage(X_scaled, method='ward') 
dendrogram(linked,
           orientation='top',
           distance_sort='descending',
           show_leaf_counts=True)
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering
```

```python
n_clusters = 3
agg_cluster = AgglomerativeClustering(n_clusters=n_clusters, linkage='ward')
cluster_labels = agg_cluster.fit_predict(X_scaled)

print(f"Cluster labels: {cluster_labels}")
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(10, 7))
plt.title(f'Hierarchical Clustering of Iris Dataset (k={n_clusters})')
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=cluster_labels, cmap='viridis', s=50)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend(handles=scatter.legend_elements()[0], labels=[f'Cluster {i}' for i in range(n_clusters)])
plt.show()
```

```python
from sklearn.metrics import adjusted_rand_score
ar_score = adjusted_rand_score(y, cluster_labels)
print(f"Adjusted Rand Score: {ar_score}")
```
