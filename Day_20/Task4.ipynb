```python
df = pd.get_dummies(df, drop_first=True)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score, root_mean_squared_error
```

```python
X = df.drop('median_house_value', axis=1) 
y = df['median_house_value'] 
```

```python
X
```

```python
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
regressor = DecisionTreeRegressor(max_depth=4, random_state=42)
```

```python
regressor.fit(X_train, y_train)
```

```python
y_pred = regressor.predict(X_test)

mse = mean_squared_error(y_test, y_pred)
print(f"Mean Squared Error: {mse:.4f}")
```

```python
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)
```

```python
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
print(f"R-squared (R²): {r2:.4f}")
print(f"Mean Squared Error (MSE): {mse:.2f}")
```

```python
train_mse_scores = []
test_mse_scores = []
train_r2_scores = []
test_r2_scores = []
depths = range(1, 21)
```

```python
for depth in depths:
    
    regressor = DecisionTreeRegressor(max_depth=depth, random_state=42)
    
    regressor.fit(X_train, y_train)
    

    y_train_pred = regressor.predict(X_train)
    y_test_pred = regressor.predict(X_test)
    y_test_pred = regressor.predict(X_test)
    

    train_mse = mean_squared_error(y_train, y_train_pred)
    test_mse = mean_squared_error(y_test, y_test_pred)
    train_r2 = r2_score(y_train, y_train_pred)
    test_r2 = r2_score(y_test, y_test_pred)
    

    train_mse_scores.append(train_mse)
    test_mse_scores.append(test_mse)
    train_r2_scores.append(train_r2)
    test_r2_scores.append(test_r2)

print("Analysis complete.")
```

```python
print(train_mse_scores) 
print(test_mse_scores)
print(train_r2_scores)
print(test_r2_scores) 
```

```python
plt.figure(figsize=(12, 6))
plt.plot(depths, train_mse_scores, label='Training MSE', marker='o', color='blue')
plt.plot(depths, test_mse_scores, label='Testing MSE', marker='o', color='red')
plt.xlabel('Maximum Depth of Decision Tree')
plt.ylabel('Mean Squared Error (MSE)')
plt.title('Effect of Tree Depth on Model Error')
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(depths, train_r2_scores, label='Training R-squared', marker='o', color='blue')
plt.plot(depths, test_r2_scores, label='Testing R-squared', marker='o', color='red')
plt.xlabel('Maximum Depth of Decision Tree')
plt.ylabel('R-squared Score')
plt.title('Effect of Tree Depth on Model R-squared Score')
plt.legend()
plt.grid(True)
plt.show()
```

```python
optimal_depth = depths[np.argmin(test_mse_scores)]
min_test_mse = np.min(test_mse_scores)

print(f"\nOptimal depth (based on minimum testing MSE): {optimal_depth}")
print(f"Minimum Testing MSE: {min_test_mse:.4f}")
```
